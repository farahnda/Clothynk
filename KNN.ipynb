```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('social_network_ads.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df['Purchased'].value_counts()
```

```python
gender = pd.get_dummies(df['Gender'], drop_first=True)
df = pd.concat([df, gender], axis=1)
```

```python
df.drop(['Gender'], axis=1, inplace=True)
```

```python
X = df[['Age', 'EstimatedSalary', 'Male']]
y = df['Purchased']
```

```python
from sklearn.preprocessing import StandardScaler # buat standarin data yg panjangnya beda
scaler = StandardScaler()
scaler.fit(X)
```

```python
StandardScaler()
```

```python
scaled_features = scaler.transform(X)
scaled_features
```

```python
df_feat = pd.DataFrame(scaled_features, columns=X.columns)
df_feat.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(scaled_features, y, test_size=0.2)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=1) # ambil 1 tetangga dekatnya aja
knn.fit(X_train, y_train) # model prediksi
```

```python
KNeighborsClassifier(n_neighbors=1) # belajar
```

```python
pred = knn.predict(X_test)  # model test prediksi
```

```python
from sklearn.metrics import classification_report, confusion_matrix
confusion_matrix(y_test, pred)
```

```python
print(classification_report(y_test, pred))
```
